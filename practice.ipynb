```python
import requests

response = requests.get("https://api.github.com/users/torvalds")

print(response.status_code)
print(type(response.text))
print(response.text[:200])
```

```python

data = response.json()

print(type(data))
print(data["login"])
print(data["public_repos"])
print(data["followers"])
```

```python
import json
print(json.dumps(data, indent=2))
```

```python
r = requests.get("https://api.github.com/users/torvalds/repos")
repos = r.json()

print(r.status_code)
print(type(repos))
print(len(repos))
print(repos[0]["name"])
```

```python
r = requests.get("https://api.github.com/users/torvalds/repos")
repos = r.json()

print(r.status_code)
print(type(repos))
print(len(repos))
print(repos[0]["name"])
```

```python
for repo in repos:
    print(repo["name"], "-", repo["stargazers_count"])
```

```python
import pandas as pd

df = pd.json_normalize(repos)
print(df.shape)
print(df[["name", "language", "stargazers_count"]].sort_values("stargazers_count", ascending=False).head())
```
